```python
# Basic KPI calculations
average_delivery_time = df["Delivery_Time_Hours"].mean()
average_cost = df["Transportation_Cost"].mean()
on_time_rate = (df["Delivery_Status"] == "On-Time").mean() * 100
```

```python
print("--- Basic Logistics KPIs ---")
print(f"Average Delivery Time: {average_delivery_time:.2f} hours")
print(f"Average Transportation Cost: {average_cost:.2f}")
print(f"On-Time Delivery Rate: {on_time_rate:.2f}%")
```
