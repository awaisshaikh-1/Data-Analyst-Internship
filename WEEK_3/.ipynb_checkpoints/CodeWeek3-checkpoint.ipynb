```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("data.csv")
```

```python
print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python
print("First 5 records:")
print(df.head())
```

```python
numeric_columns = [
    "Distance_KM",
    "Delivery_Time_Hours",
    "Transportation_Cost",
    "Shipment_Weight_KG"
]
```

```python
available_columns = [
    col for col in numeric_columns
    if col in df.columns
]

print("Statistical Summary:")
print(df[available_columns].describe())

print("Average values:")
print(df[available_columns].mean())

print("Median values:")
print(df[available_columns].median())
```

```python
plt.hist(df["Delivery_Time_Hours"].dropna(),
         bins=10, edgecolor="black")

plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time (Hours)")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.boxplot(x=df["Transportation_Cost"])

plt.title("Transportation Cost Distribution")
plt.xlabel("Transportation Cost")

plt.show()
```

```python
plt.scatter(df["Distance_KM"],
            df["Transportation_Cost"])

plt.title("Distance vs Transportation Cost")
plt.xlabel("Distance (KM)")
plt.ylabel("Transportation Cost")

plt.show()
```

```python
vehicle_time = df.groupby(
    "Vehicle_Type"
)["Delivery_Time_Hours"].mean()

vehicle_time.plot(kind="bar", figsize=(7, 4))

plt.title("Average Delivery Time by Vehicle")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Hours")
plt.show()
```

```python
plt.plot(df["Delivery_Time_Hours"].head(20),
         marker="o")

plt.title("Delivery Time for First 20 Records")
plt.xlabel("Record Number")
plt.ylabel("Delivery Time (Hours)")

plt.show()
```

```python
numeric_data = df[
    ["Distance_KM",
     "Delivery_Time_Hours",
     "Transportation_Cost",
     "Shipment_Weight_KG"]
]

sns.heatmap(numeric_data.corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Logistics Correlation Heatmap")

plt.show()
```

```python
mean_values = df[
    ["Distance_KM",
     "Delivery_Time_Hours",
     "Transportation_Cost"]
].mean()

median_values = df[
    ["Distance_KM",
     "Delivery_Time_Hours",
     "Transportation_Cost"]
].median()

comparison = pd.DataFrame({
    "Mean": mean_values,
    "Median": median_values
})

comparison.plot(kind="bar", figsize=(8, 4))

plt.title("Mean vs Median Comparison")
plt.ylabel("Value")
plt.tight_layout()

plt.show()
```

```python
std_values = df[
    ["Distance_KM",
     "Delivery_Time_Hours",
     "Transportation_Cost",
     "Shipment_Weight_KG"]
].std()

std_values.plot(kind="bar", figsize=(8, 4))

plt.title("Standard Deviation of Logistics Variables")
plt.ylabel("Standard Deviation")

plt.show()
```

```python
status_count = df["Delivery_Status"].value_counts()

status_count.plot(kind="bar")

plt.title("Shipment Delivery Status")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Shipments")

plt.show()
```

```python
print("Mean:")
print(df["Delivery_Time_Hours"].mean())

print("\nMedian:")
print(df["Delivery_Time_Hours"].median())

print("\nStandard Deviation:")
print(df["Delivery_Time_Hours"].std())

print("\nMinimum:")
print(df["Delivery_Time_Hours"].min())

print("\nMaximum:")
print(df["Delivery_Time_Hours"].max())

print("\nVariance:")
print(df["Delivery_Time_Hours"].var())
```

```python
sns.countplot(
    data=df,
    x="Vehicle_Type"
)

plt.title("Shipment Count by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Number of Shipments")

plt.show()
```

```python
sns.pairplot(
    df[[
        "Distance_KM",
        "Delivery_Time_Hours",
        "Transportation_Cost",
        "Shipment_Weight_KG"
    ]].dropna()
)

plt.show()
```
